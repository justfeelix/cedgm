```python
# Adjacency clustering
from matplotlib.patches import Rectangle
from baseframework.offline_clustering.adjacency import adjacency_clustering

X, _ = datahandler.load_data("complex9")

# PHASE 1: "ONLINE" summarization (The Grid)
# ==========================================
grid_size = 0.0365

#  map all points to their grid cell indices
grid_indices = np.floor(X / grid_size).astype(int)

# count how many points fell into each unique grid cell
unique_cells, counts = np.unique(grid_indices, axis=0, return_counts=True)

# calculate the actual X/Y center coordinates of these populated grid cells
summary_centers = unique_cells * grid_size + (grid_size / 2)

# find the "status" of each summary based on its density (count)
density_threshold = 4
# if a cell has 10 or more points, it's 'dense'. Otherwise, we consider it 'noise'.
status = np.array(["dense" if c >= density_threshold else "noise" for c in counts])


# PHASE 2: "OFFLINE" adjacency clustering
# ==========================================
# We want cells that are directly next to each other to connect.
# The distance between two adjacent cell centers is exactly 'grid_size'.
# The adjacency.py script connects points if distance <= sim_thresh[i] + sim_thresh[j]
# So we set the individual threshold to slightly more than half the grid size
sim_thresh = (grid_size / 2) * 1.1


labels = adjacency_clustering(
    X=summary_centers, 
    metric='euclidean', 
    status=status, 
    similarity_thresholds=sim_thresh
)

# number of unique clusters found (ignoring noise label -1)
num_clusters = len(np.unique(labels[labels != -1]))
print(f"Algorithm found {num_clusters} distinct clusters!")

# PHASE 3: VISUALIZATION
# ==========================================
fig, ax = plt.subplots(figsize=(10, 8))

# plot the raw deleted data points in the background
ax.scatter(X[:, 0], X[:, 1], c='lightgrey', s=5, alpha=0.3, label="Raw Data (Deleted)")

cmap = plt.cm.get_cmap('tab20', len(np.unique(labels)))

for i, center in enumerate(summary_centers):
    cluster_id = labels[i]
    if cluster_id == -1:
        continue # skip noise cells
        
    # calculate bottom-left corner for the rectangle drawing
    bottom_left = center - (grid_size / 2)
    
    # draw the grid cell summary
    rect = Rectangle(bottom_left, grid_size, grid_size, 
                     linewidth=1, edgecolor='black', 
                     facecolor=cmap(cluster_id), alpha=0.8)
    ax.add_patch(rect)

ax.set_title("Adjacency Clustering on Grid Summaries")
ax.set_aspect('equal', 'box')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.lines import Line2D

def plot_cedgm_state(X_raw, cmcs, omcs, edges, macro_clusters, cell_size=0.5):
    """
    Visualizes the CEDGM Graph Topology, Grid, and Outliers.
    X_raw: numpy array of raw data points (for background context)
    cmcs: dict of {cmc_id: {'center': [x, y], 'radius': r}}
    omcs: dict of {omc_id: {'center': [x, y], 'radius': r}}
    edges: dict of {cmc_id: [connected_cmc_ids]} (Adjacency List)
    macro_clusters: dict of {cmc_id: macro_cluster_id}
    cell_size: float, the physical width of the grid squares
    """
    fig, ax = plt.subplots(figsize=(12, 10))

    # 1. Draw the Grid Partitioning (Spatial Index Layer)
    if cell_size > 0:
        # Find boundaries of the data
        x_min, x_max = np.min(X_raw[:, 0]), np.max(X_raw[:, 0])
        y_min, y_max = np.min(X_raw[:, 1]), np.max(X_raw[:, 1])

        # Create tick marks spaced exactly by cell_size
        x_ticks = np.arange(np.floor(x_min) - cell_size, np.ceil(x_max) + cell_size, cell_size)
        y_ticks = np.arange(np.floor(y_min) - cell_size, np.ceil(y_max) + cell_size, cell_size)

        ax.set_xticks(x_ticks)
        ax.set_yticks(y_ticks)
        # Hide the tick labels to keep it clean, but keep the grid lines
        ax.set_xticklabels([])
        ax.set_yticklabels([])

        # Draw the actual grid lines
        ax.grid(True, which='both', color='lightgrey', linestyle='--', linewidth=0.8, zorder=0)

    # 2. Plot raw data faintly in the background
    ax.scatter(X_raw[:, 0], X_raw[:, 1], c='gainsboro', s=10, alpha=0.5, label='Raw Data Stream', zorder=1)

    # 3. Draw Outliers (OMCs - The Grey Bubbles)
    for omc_id, omc in omcs.items():
        center = omc['center']
        radius = omc['radius']

        # Faint grey bubble
        circle = Circle(center, radius, color='grey', alpha=0.2, zorder=2)
        ax.add_patch(circle)
        # Small 'x' for outlier center
        ax.scatter(center[0], center[1], color='dimgrey', s=15, marker='x', zorder=3)

    cmap = plt.get_cmap('tab10') # Color map for Macro-Clusters
    drawn_edges = set()

    # 4. Draw Edges (Links between CMCs)
    for cmc_id, neighbors in edges.items():
        if cmc_id not in cmcs: continue # Safety check
        c1 = cmcs[cmc_id]['center']
        mc_id = macro_clusters.get(cmc_id, 0)
        color = cmap(mc_id % 10)

        for neighbor_id in neighbors:
            if neighbor_id not in cmcs: continue

            # Avoid drawing the same bidirectional line twice
            edge_tuple = tuple(sorted((cmc_id, neighbor_id)))
            if edge_tuple not in drawn_edges:
                c2 = cmcs[neighbor_id]['center']
                # Draw thick line connecting centers
                ax.plot([c1[0], c2[0]], [c1[1], c2[1]], color=color, linewidth=2.5, zorder=4)
                drawn_edges.add(edge_tuple)

    # 5. Draw Core Micro-Clusters (CMCs)
    for cmc_id, cmc in cmcs.items():
        center = cmc['center']
        radius = cmc['radius']
        mc_id = macro_clusters.get(cmc_id, 0)
        color = cmap(mc_id % 10)

        # Draw the Kernel (Bubble)
        circle = Circle(center, radius, color=color, alpha=0.5, zorder=5)
        ax.add_patch(circle)

        # Draw the Center Dot
        ax.scatter(center[0], center[1], color='black', s=20, zorder=6)

    ax.set_title(f"CEDGM Live Graph Topology (Grid Size: {cell_size})", fontsize=14)
    ax.set_aspect('equal', 'box')

    # Custom legend
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', label='Raw Data', markerfacecolor='gainsboro', markersize=8),
        Line2D([0], [0], marker='o', color='w', label='Outlier (OMC)', markerfacecolor='grey', alpha=0.3, markersize=12),
        Line2D([0], [0], marker='o', color='w', label='Core (CMC)', markerfacecolor=cmap(0), alpha=0.5, markersize=12),
        Line2D([0], [0], color=cmap(0), lw=2.5, label='Graph Edge'),
        Line2D([0], [0], color='lightgrey', lw=1.5, linestyle='--', label='Grid Partition')
    ]
    ax.legend(handles=legend_elements, loc='upper right', bbox_to_anchor=(1.25, 1))
    plt.tight_layout()
    plt.show()

# ==========================================
# MOCK SIMULATION (To test the visualizer)
# ==========================================
if __name__ == "__main__":
    # Generate fake raw data
    X_shape1 = np.random.normal(loc=[2, 2], scale=0.5, size=(200, 2))
    X_shape2 = np.random.normal(loc=[5, 4], scale=0.6, size=(200, 2))
    X_noise = np.random.uniform(low=0, high=7, size=(30, 2)) # Random scatter
    X_mock = np.vstack([X_shape1, X_shape2, X_noise])

    # Simulate Anton's CMCs
    mock_cmcs = {
        1: {'center': [1.8, 1.8], 'radius': 0.35},
        2: {'center': [2.2, 2.0], 'radius': 0.4},
        3: {'center': [1.9, 2.4], 'radius': 0.35},
        4: {'center': [4.8, 3.8], 'radius': 0.4},
        5: {'center': [5.2, 4.2], 'radius': 0.35},
    }

    # Simulate Anton's OMCs (Notice they are isolated)
    mock_omcs = {
        101: {'center': [1.0, 4.5], 'radius': 0.2},
        102: {'center': [6.5, 1.5], 'radius': 0.25},
        103: {'center': [3.5, 3.0], 'radius': 0.15}, # Sitting between the clusters!
    }

    # Simulate Paul's Adjacency Graph (1,2,3 touch. 4,5 touch).
    mock_edges = {
        1: [2],
        2: [1, 3],
        3: [2],
        4: [5],
        5: [4]
    }

    # Simulate Paul's Connected Components Output
    mock_macro_clusters = {
        1: 100,
        2: 100,
        3: 100,
        4: 101,
        5: 101
    }

    # Render with a grid cell size of 1.0!
    plot_cedgm_state(
        X_raw=X_mock,
        cmcs=mock_cmcs,
        omcs=mock_omcs,
        edges=mock_edges,
        macro_clusters=mock_macro_clusters,
        cell_size=0.5
    )
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
from matplotlib.lines import Line2D
from ipywidgets import interact, IntSlider

def plot_cedgm_state(X_raw, cmcs, omcs, edges, macro_clusters, cell_size=0.5, batch_name="Current"):
    """
    Renders a single frame (batch) of the CEDGM state.
    """
    fig, ax = plt.subplots(figsize=(10, 8))

    # 1. Draw the Grid Partitioning
    if cell_size > 0 and len(X_raw) > 0:
        x_min, x_max = np.min(X_raw[:, 0]), np.max(X_raw[:, 0])
        y_min, y_max = np.min(X_raw[:, 1]), np.max(X_raw[:, 1])
        x_ticks = np.arange(np.floor(x_min) - cell_size, np.ceil(x_max) + cell_size, cell_size)
        y_ticks = np.arange(np.floor(y_min) - cell_size, np.ceil(y_max) + cell_size, cell_size)

        ax.set_xticks(x_ticks)
        ax.set_yticks(y_ticks)
        ax.set_xticklabels([])
        ax.set_yticklabels([])
        ax.grid(True, which='both', color='lightgrey', linestyle='--', linewidth=0.8, zorder=0)

    # 2. Plot cumulative raw data up to this batch
    if len(X_raw) > 0:
        ax.scatter(X_raw[:, 0], X_raw[:, 1], c='gainsboro', s=10, alpha=0.5, label='Raw Data Stream', zorder=1)

    # 3. Draw Outliers (OMCs) - Now accurately centered on sparse data points
    for omc_id, omc in omcs.items():
        center = omc['center']
        radius = omc['radius']
        circle = Circle(center, radius, color='grey', alpha=0.2, zorder=2)
        ax.add_patch(circle)
        ax.scatter(center[0], center[1], color='dimgrey', s=15, marker='x', zorder=3)

    cmap = plt.get_cmap('tab10')
    drawn_edges = set()

    # 4. Draw Edges
    for cmc_id, neighbors in edges.items():
        if cmc_id not in cmcs: continue
        c1 = cmcs[cmc_id]['center']
        mc_id = macro_clusters.get(cmc_id, 0)
        color = cmap(mc_id % 10)

        for neighbor_id in neighbors:
            if neighbor_id not in cmcs: continue
            edge_tuple = tuple(sorted((cmc_id, neighbor_id)))
            if edge_tuple not in drawn_edges:
                c2 = cmcs[neighbor_id]['center']
                ax.plot([c1[0], c2[0]], [c1[1], c2[1]], color=color, linewidth=2.5, zorder=4)
                drawn_edges.add(edge_tuple)

    # 5. Draw CMCs
    for cmc_id, cmc in cmcs.items():
        center = cmc['center']
        radius = cmc['radius']
        mc_id = macro_clusters.get(cmc_id, 0)
        color = cmap(mc_id % 10)

        circle = Circle(center, radius, color=color, alpha=0.5, zorder=5)
        ax.add_patch(circle)
        ax.scatter(center[0], center[1], color='black', s=20, zorder=6)

    ax.set_title(f"CEDGM Stream Live View - {batch_name}", fontsize=14)
    ax.set_aspect('equal', 'box')

    # Static limits so the camera doesn't jump around during the stream
    ax.set_xlim(0, 8)
    ax.set_ylim(0, 8)

    legend_elements = [
        Line2D([0], [0], marker='o', color='w', label='Raw Data', markerfacecolor='gainsboro', markersize=8),
        Line2D([0], [0], marker='x', color='w', label='Outlier (OMC)', markeredgecolor='dimgrey', markersize=8),
        Line2D([0], [0], marker='o', color='w', label='Core (CMC)', markerfacecolor=cmap(0), alpha=0.5, markersize=12),
        Line2D([0], [0], color=cmap(0), lw=2.5, label='Graph Edge'),
    ]
    ax.legend(handles=legend_elements, loc='upper right', bbox_to_anchor=(1.25, 1))
    plt.tight_layout()
    plt.show()

# ==========================================
# MOCK STREAM GENERATOR (4 Distinct Time Steps)
# ==========================================
def generate_mock_stream():
    stream_history = []

    # Base shapes
    X_shape1 = np.random.normal(loc=[2, 2], scale=0.5, size=(100, 2))
    X_shape2 = np.random.normal(loc=[5, 4], scale=0.6, size=(100, 2))

    # BATCH 0: Early stream. Only left shape is forming. Isolated OMCs appear exactly on noise points.
    noise_pt_1 = [1.0, 5.0]
    noise_pt_2 = [6.0, 1.5]
    stream_history.append({
        'batch_name': 'Batch 0 (T=0)',
        'X_raw': np.vstack([X_shape1[:20], [noise_pt_1], [noise_pt_2]]),
        'cmcs': {1: {'center': [2.0, 2.0], 'radius': 0.3}},
        'omcs': {101: {'center': noise_pt_1, 'radius': 0.1}, 102: {'center': noise_pt_2, 'radius': 0.1}},
        'edges': {},
        'macro_clusters': {1: 100}
    })

    # BATCH 1: Left shape expands. Right shape starts as an OMC.
    stream_history.append({
        'batch_name': 'Batch 1 (T=100)',
        'X_raw': np.vstack([X_shape1[:60], X_shape2[:10], [noise_pt_1], [noise_pt_2]]),
        'cmcs': {1: {'center': [1.8, 1.8], 'radius': 0.35}, 2: {'center': [2.2, 2.2], 'radius': 0.35}},
        'omcs': {101: {'center': noise_pt_1, 'radius': 0.1}, 102: {'center': noise_pt_2, 'radius': 0.1}, 103: {'center': [5.0, 4.0], 'radius': 0.2}},
        'edges': {1: [2], 2: [1]},
        'macro_clusters': {1: 100, 2: 100}
    })

    # BATCH 2: Right shape OMC (103) promotes to CMC 3. Two distinct Macro-Clusters exist.
    stream_history.append({
        'batch_name': 'Batch 2 (T=200)',
        'X_raw': np.vstack([X_shape1, X_shape2[:50], [noise_pt_1], [noise_pt_2]]),
        'cmcs': {1: {'center': [1.8, 1.8], 'radius': 0.35}, 2: {'center': [2.2, 2.2], 'radius': 0.35}, 3: {'center': [5.0, 4.0], 'radius': 0.4}},
        'omcs': {101: {'center': noise_pt_1, 'radius': 0.1}}, # noise_pt_2 decayed and died!
        'edges': {1: [2], 2: [1]},
        'macro_clusters': {1: 100, 2: 100, 3: 101} # 3 gets a new ID (101)
    })

    # BATCH 3: A bridge forms! CMC 4 connects the two shapes. They merge into Macro-Cluster 100.
    X_bridge = np.random.normal(loc=[3.5, 3.0], scale=0.2, size=(30, 2))
    stream_history.append({
        'batch_name': 'Batch 3 (T=300) - BRIDGE MERGE',
        'X_raw': np.vstack([X_shape1, X_shape2, X_bridge, [noise_pt_1]]),
        'cmcs': {
            1: {'center': [1.8, 1.8], 'radius': 0.35}, 2: {'center': [2.5, 2.3], 'radius': 0.4},
            3: {'center': [5.0, 4.0], 'radius': 0.4},  4: {'center': [3.7, 3.2], 'radius': 0.45} # The Bridge
        },
        'omcs': {101: {'center': noise_pt_1, 'radius': 0.1}},
        'edges': {1: [2], 2: [1, 4], 4: [2, 3], 3: [4]},
        'macro_clusters': {1: 100, 2: 100, 4: 100, 3: 100} # ALL MERGED!
    })

    return stream_history

# ==========================================
# INTERACTIVE RENDERER
# ==========================================
if __name__ == "__main__":
    history = generate_mock_stream()

    def view_stream(batch_index):
        state = history[batch_index]
        plot_cedgm_state(
            X_raw=state['X_raw'],
            cmcs=state['cmcs'],
            omcs=state['omcs'],
            edges=state['edges'],
            macro_clusters=state['macro_clusters'],
            cell_size=1.0,
            batch_name=state['batch_name']
        )

    # Creates an interactive slider in Jupyter
    interact(view_stream, batch_index=IntSlider(min=0, max=len(history)-1, step=1, value=0, description='Time Step:'))
```
